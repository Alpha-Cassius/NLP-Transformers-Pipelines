## Importing Modules

```python
from transformers import pipeline
```

## 1. Initialize the Named Entity Recognition (NER) pipeline.
* By default, this loads a BERT model trained to recognize 4 categories:
* PER (Person), ORG (Organization), LOC (Location), and MISC (Miscellaneous).
* We use aggregation_strategy="simple" so that multi-word entities
* (like "Vaibhav Pandey") are grouped into a single result rather than split into sub-words.

```python
print("Loading NER model...")
ner_pipeline = pipeline("ner", aggregation_strategy="simple")
```

## 2. Define the text we want to analyze.
I've included a mix of names, places, and tech organizations.

```python
text = "Vaibhav Pandey is building an agricultural AI project called AgriWater-OS in Bihar, India. He also runs experiments using Google Colab."
```

## 3. Extract the entities.

```python
print(f"\nAnalyzing text:\n'{text}'\n")
entities = ner_pipeline(text)
```

## 4. Loop through the extracted entities and format the output.

```python
print("Extracted Entities:")
print("-" * 60)
for entity in entities:
    # We display the word, its classification type (entity_group), and the model's confidence score.
    print(f"Word: {entity['word']:<15} | Type: {entity['entity_group']:<5} | Confidence: {entity['score']:.4f}")
print("-" * 60)
```
