## Importing Modules

```python
from transformers import pipeline
```

# Part 1: Text Summarization

## 1. Initialize the text summarization pipeline.


```python
print("Loading chat model...")
# Initialize the unified text-generation pipeline with a lightweight conversational model
chat_pipeline = pipeline("text-generation", model="Qwen/Qwen1.5-0.5B-Chat")
```

## 2. Define the text sample we want to summarize.




```python
long_text = """
Artificial intelligence (AI) is transforming the agriculture industry by enabling farmers to optimize their yields,
reduce water usage, and detect crop diseases early. Machine learning models can analyze satellite imagery and
sensor data to provide actionable insights. As climate change creates unpredictable weather patterns, AI-driven
precision agriculture becomes essential for global food security. By integrating automated robotics and intelligent
irrigation systems, the modern farm is becoming a highly efficient, data-driven ecosystem.
"""
```

## 3. Format the prompt as a conversation history


```python
summary_messages = [
    {"role": "system", "content": "You are a helpful assistant that provides short, concise summaries."},
    {"role": "user", "content": f"Summarize this text in exactly one sentence:\n\n{long_text}"}
]
```

## 4. Generating summary and printing it as output

```python
print("\nGenerating Summary...")
# max_new_tokens limits the length of the generated response
summary = chat_pipeline(summary_messages, max_new_tokens=60)

print("Summary:")
print("-" * 60)
# Extract just the latest response from the chat history
print(summary[0]['generated_text'][-1]['content'])
print("-" * 60)
```

# Part 2: Translation (English to French)

## 5. Define the text sample we want to translate.


```python
text_to_translate = "Artificial intelligence is transforming agriculture and improving global food security."
```

## 6. Give the model a new persona for translation



```python
translate_messages = [
    {"role": "system", "content": "You are a professional English to French translator. Only output the French translation."},
    {"role": "user", "content": f"Translate this text to French: '{text_to_translate}'"}
]
```

## 7. Generating translation and printing it as output

```python
print("\nGenerating Translation...")
translation = chat_pipeline(translate_messages, max_new_tokens=40)

print("Translation (French):")
print("-" * 60)
print(translation[0]['generated_text'][-1]['content'])
print("-" * 60)
```
