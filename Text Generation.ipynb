## Importing Modules

```python
from transformers import pipeline
```

## 1. Initialize the text generation pipeline
We are using GPT-2, a classic foundational model for text generation.

```python
print("Loading model...")
generator = pipeline('text-generation', model='gpt2')
print("Model loaded...")
```

## 2. Define your prompt (the starting text)


```python
prompt = "The future of artificial intelligence in agriculture is"
print(f"\nPrompt: '{prompt}'\n")
```

## 3. Generate the text

```python
print("Generating response...\n")

# max_length controls how long the output is
# num_return_sequences controls how many different variations it generates
results = generator(prompt, max_length=50, num_return_sequences=1, truncation=True)
```

## 4. Display the result

```python
print(results[0]['generated_text'])
```
