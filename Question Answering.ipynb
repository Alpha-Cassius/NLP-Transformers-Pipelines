 Extracting answers out of a given text reference.

## Importing Modules

```python
from transformers import pipeline
```

## 1. Initialize the text-generation pipeline with the chat model

```python
print("Loading chat model for Question Answering...")
qa_pipeline = pipeline("text-generation", model="Qwen/Qwen1.5-0.5B-Chat")
```

## 2. Provide the reference text (context).

```python
context_text = """
The Transformer is a deep learning architecture introduced in 2017 by Google in the paper 'Attention Is All You Need'.
It relies entirely on self-attention mechanisms, completely dispensing with recurrence (RNNs) and convolutions (CNNs).
This architecture allows for highly parallelized processing, making it exceptionally fast to train on modern GPUs.
It serves as the foundation for modern large language models like GPT, BERT, and LLaMA.
"""
```

## 3. Define the questions.

```python
questions = [
    "Who introduced the Transformer architecture?",
    "What year was it introduced?",
    "What models use this as their foundation?"
]

print(f"\nContext Provided:\n{context_text}\n")
print("Extracting Answers...\n")
print("-" * 60)
```

## 4. Loop through the questions and prompt the model.

```python
for question in questions:
    # We use a strict system prompt so it only extracts facts from the context.
    messages = [
        {"role": "system", "content": "You are a reading comprehension assistant. Answer the question strictly using only the provided context. Keep the answer extremely short."},
        {"role": "user", "content": f"Context: {context_text}\n\nQuestion: {question}"}
    ]

    # Generate the answer
    result = qa_pipeline(messages, max_new_tokens=30)

    print(f"Question: {question}")
    # Extract the assistant's latest response from the chat sequence
    print(f"Answer: {result[0]['generated_text'][-1]['content']}")
    print("-" * 60)
```
