## Importing Modules

```python
from transformers import pipeline
```

## 1. Initialize the sentiment analysis pipeline.
* If you don't specify a model, Hugging Face defaults to 'distilbert-base-uncased-finetuned-sst-2-english'.
* This is a lightweight model specifically trained to classify positive and negative sentiment.

```python
print("Loading sentiment analysis model...")
sentiment_analyzer = pipeline('sentiment-analysis')
```

## 2. Define the text samples we want to analyze.

```python
reviews = [
    "I absolutely love the new agricultural AI tool, it saves me hours of work!",
    "The application keeps crashing every time I try to upload my soil data. Terrible experience.",
    "The software update is okay, it didn't change much for my daily routine."
]
```

## 3. Pass the texts through the model.

```python
print("\nAnalyzing sentiments...\n")
results = sentiment_analyzer(reviews)
```

## 4. Loop through the results and print them out with their confidence scores.

```python
for review, result in zip(reviews, results):
    print(f"Text: '{review}'")
    # The result contains a 'label' (POSITIVE/NEGATIVE) and a 'score' (probability from 0 to 1).
    print(f"Sentiment: {result['label']} (Confidence Score: {result['score']:.4f})")
```
